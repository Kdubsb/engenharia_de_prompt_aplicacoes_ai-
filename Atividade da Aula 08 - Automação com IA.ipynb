### Simulação de Arquivos Bagunçados

Vamos criar alguns arquivos de exemplo para simular um diretório bagunçado e testar a função de organização.

```python
import os
import shutil

# Lista de arquivos de exemplo (mantida do estado do kernel)
example_files = [
    'documento_importante.pdf', 'foto_ferias.jpg', 'relatorio_mensal.xlsx',
    'codigo_python.py', 'texto_simples.txt', 'apresentacao.pptx',
    'imagem_logotipo.png', 'dados_brutos.csv', 'arquivo_sem_extensao',
    'outra_foto.jpeg', 'musica.mp3', 'video.mp4', 'backup.zip', 'log.txt',
    'script.js', 'estilo.css', 'template.html', 'dados.json', 'config.ini',
    'licenca.md', 'teste.tar.gz', 'outro_doc.docx', 'planilha.ods', 'grafico.svg'
]

# Criar os arquivos de exemplo no diretório atual
for filename in example_files:
    with open(filename, 'w') as f:
        f.write(f'Conteúdo de {filename}')

print("Arquivos de exemplo criados:")
print(os.listdir('.'))

# --- Função para organizar arquivos ---
def organizar_arquivos(diretorio_base='.'):

    arquivos = [f for f in os.listdir(diretorio_base) if os.path.isfile(os.path.join(diretorio_base, f))]

    # Dicionário para mapear extensões para nomes de pastas
    diretorios_por_extensao = {
        '.pdf': 'PDF_Files',
        '.doc': 'Documentos',
        '.docx': 'Documentos',
        '.txt': 'Text_Files',
        '.jpg': 'Image_Files',
        '.jpeg': 'Image_Files',
        '.png': 'Image_Files',
        '.gif': 'Image_Files',
        '.xlsx': 'Spreadsheets',
        '.csv': 'Data_Files',
        '.py': 'Python_Scripts',
        '.js': 'Scripts',
        '.html': 'Web_Files',
        '.css': 'Web_Files',
        '.mp3': 'Audio_Files',
        '.mp4': 'Video_Files',
        '.zip': 'Archives',
        '.gz': 'Archives',
        '.tar.gz': 'Archives',
        '.pptx': 'Presentations',
        '.json': 'Data_Files',
        '.ini': 'Config_Files',
        '.md': 'Markdown_Files',
        '.ods': 'Spreadsheets',
        '.svg': 'Image_Files'
    }

    pastas_criadas = set()
    pastas_limit = 10

    print("\nIniciando organização de arquivos...")

    for arquivo in arquivos:
        if arquivo.startswith('.'):
            continue

        nome, extensao = os.path.splitext(arquivo)
        extensao = extensao.lower()

        # Tenta encontrar um diretório mapeado para a extensão
        diretorio_destino_nome = diretorios_por_extensao.get(extensao, 'Outros')

        # Se for um arquivo sem extensão, coloca em 'Outros'
        if not extensao and not diretorios_por_extensao.get(extensao):
            diretorio_destino_nome = 'Outros'

        diretorio_destino_path = os.path.join(diretorio_base, diretorio_destino_nome)

        if diretorio_destino_nome not in pastas_criadas and len(pastas_criadas) >= pastas_limit:
            diretorio_destino_path = os.path.join(diretorio_base, 'Outros_Restantes')
            if 'Outros_Restantes' not in pastas_criadas:
                os.makedirs(diretorio_destino_path, exist_ok=True)
                pastas_criadas.add('Outros_Restantes')
        elif diretorio_destino_nome not in pastas_criadas:
            os.makedirs(diretorio_destino_path, exist_ok=True)
            pastas_criadas.add(diretorio_destino_nome)

        # Verifica se o arquivo já está na pasta de destino
        if not os.path.exists(os.path.join(diretorio_destino_path, arquivo)):
            try:
                shutil.move(os.path.join(diretorio_base, arquivo), diretorio_destino_path)
                print(f"Movido: {arquivo} -> {diretorio_destino_path}")
            except Exception as e:
                print(f"Erro ao mover {arquivo}: {e}")
        else:
            print(f"Arquivo {arquivo} já existe em {diretorio_destino_path}, ignorando movimento.")

    print("\nOrganização concluída.")
    print("Pastas criadas:", sorted(list(pastas_criadas)))
    print("Conteúdo do diretório após organização:")
    print(os.listdir('.'))

# Executar a função de organização
organizar_arquivos()
```

### Verificação da Organização

Após a execução, você deve ver novas pastas criadas (ex: `PDF_Files`, `JPG_Files`, etc.), e os arquivos de exemplo devem ter sido movidos para suas respectivas pastas.

Missão 02: Consulta Simples a APIs

```python
import requests

def consultar_cep(cep):
    """
    Consulta a API do ViaCEP para obter informações de endereço de um determinado CEP.

    Args:
        cep (str): O número do CEP a ser consultado (apenas dígitos).

    Returns:
        dict or None: Um dicionário com os dados do endereço se a consulta for bem-sucedida,
                      caso contrário, None.
    """
    # Remove qualquer caractere que não seja dígito do CEP
    cep = ''.join(filter(str.isdigit, cep))

    if len(cep) != 8:
        print("CEP inválido. Por favor, digite 8 dígitos.")
        return None

    # CORREÇÃO: Alterado '/jsons/' para '/json/'
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        response = requests.get(url)
        response.raise_for_status()  # Lança uma exceção para códigos de status HTTP de erro (4xx ou 5xx)

        data = response.json()

        if "erro" in data and data["erro"]:
            print(f"Erro: CEP '{cep}' não encontrado.")
            return None
        else:
            return data
    except requests.exceptions.HTTPError as http_err:
        print(f"Erro HTTP: {http_err} (Status: {response.status_code})")
    except requests.exceptions.ConnectionError as conn_err:
        print(f"Erro de Conexão: Verifique sua conexão com a internet. {conn_err}")
    except requests.exceptions.Timeout as time_err:
        print(f"Timeout: A requisição demorou muito para responder. {time_err}")
    except requests.exceptions.RequestException as req_err:
        print(f"Erro na Requisição: {req_err}")
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")
    return None

def formatar_endereco(endereco_data):
    """
    Formata os dados de endereço de um dicionário para uma string legível.

    Args:
        endereco_data (dict): Dicionário contendo os dados do endereço.

    Returns:
        str: Uma string formatada com as informações do endereço.
    """
    if not endereco_data:
        return "Nenhuma informação de endereço para formatar."

    return (
        f"CEP: {endereco_data.get('cep', 'N/A')}\n"
        f"Logradouro: {endereco_data.get('logradouro', 'N/A')}\n"
        f"Complemento: {endereco_data.get('complemento', 'N/A')}\n"
        f"Bairro: {endereco_data.get('bairro', 'N/A')}\n"
        f"Localidade: {endereco_data.get('localidade', 'N/A')}\n"
        f"UF: {endereco_data.get('uf', 'N/A')}\n"
        f"IBGE: {endereco_data.get('ibge', 'N/A')}\n"
        f"GIA: {endereco_data.get('gia', 'N/A')}\n"
        f"DDD: {endereco_data.get('ddd', 'N/A')}\n"
        f"SIAFI: {endereco_data.get('siafi', 'N/A')}"
    )

# --- Exemplo de Uso ---
if __name__ == '__main__':
    cep_input = input("Digite o CEP para consulta (somente números): ")

    print("\n--- Realizando Consulta ---")
    dados_endereco = consultar_cep(cep_input)

    if dados_endereco:
        print("\n--- Endereço Encontrado ---")
        print(formatar_endereco(dados_endereco))
    else:
        print("\nNão foi possível obter os dados do endereço para o CEP informado.")
```


Missão 03: Sistema de Notificações



```python
import time # Importa o módulo time para funções relacionadas a tempo, como time.sleep e time.strftime
import random # Importa o módulo random para gerar números aleatórios, simulando leituras de sensor

temperatura_critica = 28.0 # Define a temperatura crítica em Celsius, acima da qual um alerta será disparado
intervalo_verificacao_segundos = 5 # Define o intervalo em segundos entre as verificações do sensor

def simular_leitura_sensor(): # Define uma função para simular a leitura de um sensor
    # Gera um número flutuante aleatório entre 20.0 e 35.0 e arredonda para duas casas decimais
    return round(random.uniform(20.0, 35.0), 2)

def verificar_condicao_critica(leitura_atual): # Define uma função para verificar se a leitura atual é crítica
    return leitura_atual > temperatura_critica # Retorna True se a leitura atual exceder a temperatura crítica, False caso contrário

def enviar_alerta(mensagem, tipo_alerta="console"): # Define uma função para enviar alertas
    # Obtém o timestamp atual formatado
    timestamp = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime())
    if tipo_alerta == "console": # Se o tipo de alerta for 'console'
        print(f"\n!!! ALERTA DE CONSOLE - {timestamp} !!!\n{mensagem}\n") # Imprime a mensagem de alerta no console
    elif tipo_alerta == "email": # Se o tipo de alerta for 'email' (simulado)
        print(f"\n--- E-MAIL SIMULADO - {timestamp} ---") # Imprime cabeçalho de e-mail simulado
        print(f"De: sistema.monitoramento@exemplo.com") # Imprime remetente do e-mail simulado
        print(f"Para: equipe.suporte@exemplo.com") # Imprime destinatário do e-mail simulado
        print(f"Assunto: ALERTA CRÍTICO: {mensagem.split(':')[0].strip()}") # Imprime o assunto do e-mail simulado, pegando a primeira parte da mensagem
        print(f"\nPrezada equipe,\n\n{mensagem}\n\nAtenciosamente,\nSistema de Monitoramento Autônomo") # Imprime o corpo do e-mail simulado
        print(f"--- FIM E-MAIL SIMULADO ---\n") # Imprime rodapé de e-mail simulado
    else: # Se o tipo de alerta for desconhecido
        print(f"Tipo de alerta desconhecido: {tipo_alerta}. Mensagem: {mensagem}") # Imprime uma mensagem de erro

def programa_monitoramento_autonomo(): # Define a função principal do programa de monitoramento
    print("Iniciando monitoramento autônomo...") # Informa que o monitoramento está iniciando
    estado_critico_anterior = False # Inicializa uma flag para rastrear se a condição era crítica na verificação anterior

    try: # Inicia um bloco try para capturar exceções
        while True: # Loop infinito para monitoramento contínuo
            leitura = simular_leitura_sensor() # Obtém uma nova leitura simulada do sensor
            estado_critico_atual = verificar_condicao_critica(leitura) # Verifica se a leitura atual é crítica

            print(f"[{time.strftime('%H:%M:%S')}] Leitura atual: {leitura}°C (Crítico: {estado_critico_atual})") # Imprime a leitura atual e seu status

            # Se a condição atual for crítica E a condição anterior NÃO era crítica, dispara o alerta
            if estado_critico_atual and not estado_critico_anterior:
                mensagem_alerta = f"Temperatura {leitura}°C excedeu o limite crítico de {temperatura_critica}°C!" # Monta a mensagem de alerta
                enviar_alerta(mensagem_alerta, tipo_alerta="console") # Envia o alerta para o console
                enviar_alerta(mensagem_alerta, tipo_alerta="email") # Envia o alerta como e-mail simulado

            estado_critico_anterior = estado_critico_atual # Atualiza o estado crítico anterior para a próxima iteração

            time.sleep(intervalo_verificacao_segundos) # Pausa o programa pelo intervalo definido

    except KeyboardInterrupt: # Captura a exceção quando o usuário interrompe o programa (ex: Ctrl+C)
        print("\nMonitoramento autônomo interrompido pelo usuário.") # Informa que o monitoramento foi interrompido
    except Exception as e: # Captura quaisquer outras exceções inesperadas
        print(f"Ocorreu um erro inesperado: {e}") # Imprime a mensagem de erro

if __name__ == '__main__': # Verifica se o script está sendo executado diretamente
    programa_monitoramento_autonomo() # Chama a função principal para iniciar o monitoramento
```
