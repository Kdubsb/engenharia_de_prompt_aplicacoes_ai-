# Soma Lista

```python
numeros = [10, 20, 30, 40, 50] # Inicializa uma lista de números inteiros.
soma = 0 # Inicializa uma variável 'soma' com valor zero para acumular a soma dos números.
i = 0 # Inicializa um contador 'i' com zero para ser usado no loop.
while i < len(numeros): # Inicia um loop 'while' que continua enquanto 'i' for menor que o número de elementos na lista 'numeros'.
  soma += numeros[i] # Adiciona o número atual da lista (na posição 'i') à variável 'soma'.
  i += 1 # Incrementa o contador 'i' em 1 para passar para o próximo número da lista.
print(f"A soma dos números na lista é: {soma}") # Imprime a soma total dos números da lista usando uma f-string para formatar a saída.
```



#Número Primo


```python
import math # Importa o módulo math para usar a função sqrt (raiz quadrada).

n=float(input("digite um numero: "))
def eh_primo(n): # Define uma função chamada 'eh_primo' que recebe um número inteiro 'n' como entrada.
    if n < 2: # Verifica se o número é menor que 2. Números menores que 2 não são primos.
        return False # Retorna False se 'n' for menor que 2.
    if n == 2: # Verifica se o número é 2. O número 2 é o único número primo par.
        return True # Retorna True se 'n' for 2.
    if n % 2 == 0: # Verifica se o número é par (e maior que 2). Números pares maiores que 2 não são primos.
        return False # Retorna False se 'n' for par e maior que 2.
    # Inicia um loop para verificar divisores ímpares. O loop vai de 3 até a raiz quadrada de 'n', incrementando de 2 em 2.
    # Não precisamos verificar divisores pares, pois já tratamos números pares, e não precisamos ir além da raiz quadrada de 'n'.
    for i in range(3, int(math.sqrt(n)) + 1, 2):
        if n % i == 0: # Se 'n' for divisível por 'i' (sem deixar resto), então 'n' não é primo.
            return False # Retorna False se um divisor for encontrado.
    return True # Se nenhum divisor for encontrado após todas as verificações, o número é primo e retorna True.
print(eh_primo(n))
```

# Mini Projeto

Calculadora simples

```python
def calculadora(): # Define uma função chamada 'calculadora' para encapsular a lógica da aplicação.
    print("Bem-vindo à Calculadora Simples!") # Exibe uma mensagem de boas-vindas ao usuário.
    try: # Inicia um bloco try-except para tratar possíveis erros na entrada do usuário.
        num1 = float(input("Digite o primeiro número: ")) # Solicita e converte o primeiro número para float.
        num2 = float(input("Digite o segundo número: ")) # Solicita e converte o segundo número para float.
    except ValueError: # Captura o erro se a entrada não puder ser convertida para número.
        print("Entrada inválida. Por favor, digite números.") # Informa o usuário sobre a entrada inválida.
        return # Encerra a execução da função se houver erro.

    operador = input("Escolha uma operação (+, -, *, /): ") # Solicita ao usuário que escolha uma operação.

    if operador == '+': # Verifica se o operador escolhido é adição.
        resultado = num1 + num2 # Realiza a adição dos dois números.
    elif operador == '-': # Verifica se o operador escolhido é subtração.
        resultado = num1 - num2 # Realiza a subtração dos dois números.
    elif operador == '*': # Verifica se o operador escolhido é multiplicação.
        resultado = num1 * num2 # Realiza a multiplicação dos dois números.
    elif operador == '/': # Verifica se o operador escolhido é divisão.
        if num2 == 0: # Verifica se o segundo número é zero para evitar divisão por zero.
            print("Erro: Divisão por zero não é permitida.") # Informa o erro de divisão por zero.
            return # Encerra a execução da função.
        resultado = num1 / num2 # Realiza a divisão dos dois números.
    else: # Se nenhum dos operadores válidos foi escolhido.
        print("Operador inválido. Por favor, use +, -, *, ou /.") # Informa que o operador é inválido.
        return # Encerra a execução da função.

    print(f"O resultado da operação {num1} {operador} {num2} é: {resultado}") # Exibe o resultado da operação formatado.

calculadora() # Chama a função calculadora para iniciar a aplicação.
```
