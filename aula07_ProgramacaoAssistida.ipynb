Missão 02: Consulta Simples a APIs

```python
import requests

def consultar_cep(cep):
    """
    Consulta a API do ViaCEP para obter informações de endereço de um determinado CEP.

    Args:
        cep (str): O número do CEP a ser consultado (apenas dígitos).

    Returns:
        dict or None: Um dicionário com os dados do endereço se a consulta for bem-sucedida,
                      caso contrário, None.
    """
    # Remove qualquer caractere que não seja dígito do CEP
    cep = ''.join(filter(str.isdigit, cep))

    if len(cep) != 8:
        print("CEP inválido. Por favor, digite 8 dígitos.")
        return None

    # CORREÇÃO: Alterado '/jsons/' para '/json/'
    url = f"https://viacep.com.br/ws/{cep}/json/"

    try:
        response = requests.get(url)
        response.raise_for_status()  # Lança uma exceção para códigos de status HTTP de erro (4xx ou 5xx)

        data = response.json()

        if "erro" in data and data["erro"]:
            print(f"Erro: CEP '{cep}' não encontrado.")
            return None
        else:
            return data
    except requests.exceptions.HTTPError as http_err:
        print(f"Erro HTTP: {http_err} (Status: {response.status_code})")
    except requests.exceptions.ConnectionError as conn_err:
        print(f"Erro de Conexão: Verifique sua conexão com a internet. {conn_err}")
    except requests.exceptions.Timeout as time_err:
        print(f"Timeout: A requisição demorou muito para responder. {time_err}")
    except requests.exceptions.RequestException as req_err:
        print(f"Erro na Requisição: {req_err}")
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")
    return None

def formatar_endereco(endereco_data):
    """
    Formata os dados de endereço de um dicionário para uma string legível.

    Args:
        endereco_data (dict): Dicionário contendo os dados do endereço.

    Returns:
        str: Uma string formatada com as informações do endereço.
    """
    if not endereco_data:
        return "Nenhuma informação de endereço para formatar."

    return (
        f"CEP: {endereco_data.get('cep', 'N/A')}\n"
        f"Logradouro: {endereco_data.get('logradouro', 'N/A')}\n"
        f"Complemento: {endereco_data.get('complemento', 'N/A')}\n"
        f"Bairro: {endereco_data.get('bairro', 'N/A')}\n"
        f"Localidade: {endereco_data.get('localidade', 'N/A')}\n"
        f"UF: {endereco_data.get('uf', 'N/A')}\n"
        f"IBGE: {endereco_data.get('ibge', 'N/A')}\n"
        f"GIA: {endereco_data.get('gia', 'N/A')}\n"
        f"DDD: {endereco_data.get('ddd', 'N/A')}\n"
        f"SIAFI: {endereco_data.get('siafi', 'N/A')}"
    )

# --- Exemplo de Uso ---
if __name__ == '__main__':
    cep_input = input("Digite o CEP para consulta (somente números): ")

    print("\n--- Realizando Consulta ---")
    dados_endereco = consultar_cep(cep_input)

    if dados_endereco:
        print("\n--- Endereço Encontrado ---")
        print(formatar_endereco(dados_endereco))
    else:
        print("\nNão foi possível obter os dados do endereço para o CEP informado.")
```


Missão 03: Sistema de Notificações



O Problema (briefing)

```python
import time
import random

# --- Configurações do Sistema de Alerta ---
temperatura_critica = 28.0  # Temperatura em Celsius a partir da qual consideramos a condição crítica
intervalo_verificacao_segundos = 5 # Intervalo entre as verificações (em segundos)

# --- Função para simular a leitura de um sensor (ex: temperatura) ---
def simular_leitura_sensor():
    # Simula uma leitura de temperatura entre 20.0 e 35.0 graus Celsius
    return round(random.uniform(20.0, 35.0), 2)

# --- Função para verificar se a condição é crítica ---
def verificar_condicao_critica(leitura_atual):
    return leitura_atual > temperatura_critica

# --- Função para enviar alertas (console ou e-mail simulado) ---
def enviar_alerta(mensagem, tipo_alerta="console"):
    timestamp = time.strftime("%Y-%m-%d %H:%M:%S", time.localtime())
    if tipo_alerta == "console":
        print(f"\n!!! ALERTA DE CONSOLE - {timestamp} !!!\n{mensagem}\n")
    elif tipo_alerta == "email":
        print(f"\n--- E-MAIL SIMULADO - {timestamp} ---")
        print(f"De: sistema.monitoramento@exemplo.com")
        print(f"Para: equipe.suporte@exemplo.com")
        print(f"Assunto: ALERTA CRÍTICO: {mensagem.split(':')[0].strip()}")
        print(f"\nPrezada equipe,\n\n{mensagem}\n\nAtenciosamente,\nSistema de Monitoramento Autônomo")
        print(f"--- FIM E-MAIL SIMULADO ---\n")
    else:
        print(f"Tipo de alerta desconhecido: {tipo_alerta}. Mensagem: {mensagem}")

# --- Programa Autônomo de Monitoramento ---
def programa_monitoramento_autonomo():
    print("Iniciando monitoramento autônomo...")
    estado_critico_anterior = False # Variável para rastrear o estado crítico anterior

    try:
        while True:
            leitura = simular_leitura_sensor()
            estado_critico_atual = verificar_condicao_critica(leitura)

            print(f"[{time.strftime('%H:%M:%S')}] Leitura atual: {leitura}°C (Crítico: {estado_critico_atual})")

            # Dispara alerta SOMENTE se a condição crítica for atingida AGORA e não estava crítica antes
            if estado_critico_atual and not estado_critico_anterior:
                mensagem_alerta = f"Temperatura {leitura}°C excedeu o limite crítico de {temperatura_critica}°C!"
                enviar_alerta(mensagem_alerta, tipo_alerta="console") # Alerta no console
                enviar_alerta(mensagem_alerta, tipo_alerta="email")   # E-mail simulado

            estado_critico_anterior = estado_critico_atual # Atualiza o estado anterior

            time.sleep(intervalo_verificacao_segundos)

    except KeyboardInterrupt:
        print("\nMonitoramento autônomo interrompido pelo usuário.")
    except Exception as e:
        print(f"Ocorreu um erro inesperado: {e}")

# --- Executar o programa ---
if __name__ == '__main__':
    programa_monitoramento_autonomo()
```
